```python
# ==============================================================================
# KAGGLE CELL 1: ENVIRONMENT CONFIGURATION & SECURITY REGISTRY
# ==============================================================================
!pip install fastapi uvicorn sentence-transformers numpy cryptography pydantic nest-asyncio firebase-admin -q
!npm install -g localtunnel -q
!pip install --upgrade google-genai cryptography

import os
import base64
import json
import nest_asyncio
from cryptography.hazmat.primitives.ciphers.aead import AESGCM

nest_asyncio.apply()

class Settings:
    # ─── UPDATED TO MATCH YOUR MANUALLY MOUNTED CONTAINER PATH LAYOUTS ───
    DB_PATH = "/kaggle/input/movie-graph-rag/movie_graph_rag.db"
    MOCK_DB_PATH = "movie_graph_rag_fallback.db"
    MODEL_NAME = "all-MiniLM-L6-v2"
    SERVICE_ACCOUNT_JSON_PATH = "/kaggle/input/datasets/tejasmali1234/firebase-keys/serviceAccountKey.json"

class MeshCredentialRegistry:
    @staticmethod
    def enroll_node(client_id: str, tier: str = "Premium VIP Node") -> dict:
        raw_key = AESGCM.generate_key(bit_length=256)
        b64_key = base64.b64encode(raw_key).decode('utf-8')
        token = f"CINE_MESH_{base64.b64encode(os.urandom(8)).decode('utf-8')[:10].upper()}"
        return {
            "client_id": client_id, "access_token": token,
            "allowed_scopes": ["vibe", "plot", "quotes"], "tier": tier, "secret": b64_key
        }

CREDS = MeshCredentialRegistry.enroll_node("client_node_tejas_01")
DECENTRALIZED_ACL = {
    CREDS["client_id"]: {
        "access_token": CREDS["access_token"], "allowed_scopes": CREDS["allowed_scopes"],
        "secret_bytes": base64.b64decode(CREDS["secret"])
    }
}
print(f"[OK] Cryptographic registry provisioned safely for Client: {CREDS['client_id']}")

# ==============================================================================
# KAGGLE CELL 2 (PART 1 OF 2): DIRECT DATASET PATH SYNC LAYER
# ==============================================================================
import sqlite3
import os
import shutil

class PersonalizedVectorSearchEngine:
    def __init__(self, db_file: str):
        # ─── TARGETED DATASET BASE CORES ───
        dataset_base_path = "/kaggle/input/datasets/tejasmali1234/tmdb-dataset"
        detected_db = None
        
        # Scan your explicitly specified input folder for your pre-made database
        if os.path.exists(dataset_base_path):
            for root, dirs, files in os.walk(dataset_base_path):
                for f in files:
                    if f.endswith('.db') or f.endswith('.sqlite') or f.endswith('.fallback'):
                        detected_db = os.path.join(root, f)
                        break
        
        # Fallback routing patterns if direct tree traversal fails
        self.source_db = detected_db if detected_db else db_file
        self.db_file = "/kaggle/working/movie_graph_recovered.db"
        
        print(f"[ENGINE INITIALIZATION] Linking onto pre-processed TMDB database: {self.source_db}")
        self._execute_live_database_salvage_and_repair()

    def _execute_live_database_salvage_and_repair(self):
        """Validates, links, and deploys your static dataset into active writable disk space."""
        if os.path.exists(self.db_file) and os.path.getsize(self.db_file) > 50000:
            print(f"📡 [HEALTH STATUS] Production database cache index verified active at: {self.db_file}")
            return

        print("🔧 [DATABASE MOUNTING] Instantiating high-speed file deployment...")
        
        # Copy the pre-compiled file directly into your writable working space for quick querying
        if self.source_db and os.path.exists(self.source_db) and self.source_db != self.db_file:
            try:
                shutil.copy2(self.source_db, self.db_file)
                print(f"🚀 [SUCCESS] Your TMDB database has been successfully mounted to runtime: {self.db_file}")
            except Exception as e:
                print(f"⚠️ Direct OS file copy blocked ({e}). Running table-to-table stream fallback...")
                self._migrate_rows_manually()
        else:
            if os.path.exists(self.db_file):
                print(f"✅ Active database file verified directly inside local workspace path.")
            else:
                print(f"❌ [CRITICAL ERROR] No pre-processed .db file found at path: {self.source_db}")
                print("Please check your Kaggle Dataset input structure settings tab.")

    def _migrate_rows_manually(self):
        """Streaming fallback layer to pipe database tables if direct file system locks occur."""
        dest_conn = sqlite3.connect(self.db_file)
        dest_c = dest_conn.cursor()
        
        dest_c.execute("""CREATE TABLE IF NOT EXISTS movies (
            id INT PRIMARY KEY, title TEXT, genres TEXT, overview TEXT, tagline TEXT, 
            vote_average REAL, runtime INT, homepage TEXT, director TEXT, cast_members TEXT)""")
        dest_c.execute("CREATE TABLE IF NOT EXISTS plot_vectors (movie_id INT, vector BLOB)")
        dest_c.execute("CREATE TABLE IF NOT EXISTS vibe_vectors (movie_id INT, vector BLOB)")
        dest_c.execute("CREATE TABLE IF NOT EXISTS quotes_vectors (movie_id INT, vector BLOB)")
        dest_conn.commit()

        try:
            src_conn = sqlite3.connect(self.source_db)
            src_c = src_conn.cursor()
            
            # Pipe textual relational features
            src_c.execute("SELECT id, title, genres, overview, tagline, vote_average, runtime, homepage, director, cast_members FROM movies")
            dest_c.executemany("INSERT OR IGNORE INTO movies VALUES (?,?,?,?,?,?,?,?,?,?)", src_c.fetchall())
            
            # Pipe multi-vector blobs
            for table in ["plot", "vibe", "quotes"]:
                src_c.execute(f"SELECT movie_id, vector FROM {table}_vectors")
                dest_c.executemany(f"INSERT OR IGNORE INTO {table}_vectors VALUES (?,?)", src_c.fetchall())
                
            dest_conn.commit()
            src_conn.close()
            print("✅ [STREAMING SUCCESS] Tables and vector blobs mapped over successfully.")
        except Exception as err:
            print(f"❌ Structural database streaming loop failed: {err}")
        finally:
            dest_conn.close()
        
# ==============================================================================
# KAGGLE CELL 2 (PART 2 OF 2): BATCHED RELATIONAL JOIN OPTIMIZATION PIPELINE
# ==============================================================================
import sqlite3
import numpy as np
import json

def process_personalized_recommendations(self, model, query: str, scope: str, fav_genres: list, liked_movie_ids: list) -> list:
    # 1. Instantly capture conversational intent matrix on GPU device arrays
    qv = model.encode(query, convert_to_numpy=True).astype(np.float32)
    results = []
    
    with sqlite3.connect(self.db_file) as conn:
        conn.row_factory = sqlite3.Row
        c = conn.cursor()
        
        print("📡 [HIGH-SPEED ACCELERATION] Extracting base records from disk...")
        try:
            c.execute("SELECT movie_id, title, overview, tagline, vote_average, runtime, homepage FROM movies")
            backup_rows = c.fetchall()
        except Exception as read_err:
            print(f"❌ Core table extraction constraint error: {read_err}")
            return []
            
        if not backup_rows: 
            return []

        # ─── EXTRACTION ACCELERATION LAYER: BATCH READ RELATIONAL TABLES ───
        print("⚡ Ingesting mass-relational attributes directly into local RAM caches...")
        
        # Batch Extract Genres (Groups slash strings directly by key identifiers)
        c.execute("SELECT movie_id, genre_name FROM movie_genres")
        genres_cache = {}
        for r in c.fetchall():
            mid = r["movie_id"]
            if mid not in genres_cache: genres_cache[mid] = []
            if r["genre_name"]: genres_cache[mid].append(r["genre_name"])

        # Batch Extract Top 10 Actors (Honors sequencing without looping)
        c.execute("SELECT movie_id, name FROM movie_cast WHERE order_index <= 10")
        cast_cache = {}
        for r in c.fetchall():
            mid = r["movie_id"]
            if mid not in cast_cache: cast_cache[mid] = []
            if r["name"]: cast_cache[mid].append(r["name"])

        # Batch Extract Directors (Filters job profiles directly)
        c.execute("SELECT movie_id, name FROM movie_crew WHERE job = 'Director'")
        director_cache = {r["movie_id"]: r["name"] for r in c.fetchall() if r["name"]}

        # ─── RAM DATA EXTRACTION MATRICES LOOP ───
        m_ids = []
        text_corpus = []
        movie_metadata_cache = {}
        
        for r in backup_rows:
            current_id = int(r["movie_id"])
            m_ids.append(current_id)
            
            # Map items instantly out of memory caches (0.00 milliseconds lookup time)
            genres_list = genres_cache.get(current_id, [])
            genres_joined = "/".join(genres_list)
            cast_list = cast_cache.get(current_id, [])
            director_name = director_cache.get(current_id, "Unknown Director")
            
            movie_metadata_cache[current_id] = {
                "row": r,
                "genres": genres_list,
                "genres_str": genres_joined,
                "cast": cast_list,
                "director": director_name
            }
            
            # Route text components to match search parameters
            if scope == "plot":
                text_node = str(r["overview"]) if r["overview"] else str(r["title"])
            else:
                text_node = f"Title: {r['title']}. Genres: {genres_joined}. Tagline: {r['tagline']}."
                
            text_corpus.append(text_node)
            
        print(f"🧠 Encoding {len(text_corpus)} TMDB contextual vector text matrices onto GPU space...")
        # Batched tensor calculations execute in parallel over the model layers
        matrix = model.encode(text_corpus, convert_to_numpy=True, batch_size=256, show_progress_bar=False).astype(np.float32)
        
        # ─── HIGH SPEED NUMPY DISTANCE SEARCH ───
        dots = np.dot(matrix, qv)
        norms = np.linalg.norm(matrix, axis=1) * (np.linalg.norm(qv) or 1e-9)
        norms[norms == 0] = 1e-9
        base_scores = dots / norms
        
        candidates = []
        for score, m_id in zip(base_scores, m_ids):
            meta = movie_metadata_cache[m_id]
            final_score = float(score)
            genres_lower = meta["genres_str"].lower()
            
            # Match user preference triggers
            q_low = query.lower()
            if "bollywood" in q_low or "hindi" in q_low:
                if "bollywood" in genres_lower or "hindi" in genres_lower:
                    final_score += 0.80 
                    
            for fav in fav_genres:
                if fav.lower() in genres_lower:
                    final_score += 0.25 
                    
            if m_id in liked_movie_ids:
                final_score -= 0.05  
                
            candidates.append((final_score, m_id))
        
        candidates.sort(key=lambda x: x[0], reverse=True) 
        
        # ─── PACKAGING ENCRYPTED OUTBOUND PAYLOAD ───
        for score, m_id in candidates[:10]:
            meta = movie_metadata_cache[m_id]
            row = meta["row"]
            
            results.append({
                "id": int(m_id),
                "title": str(row["title"]) if row["title"] else "Unknown Movie",
                "genres": meta["genres"] if meta["genres"] else ["Multi-Genre"],
                "description": str(row["overview"]) if row["overview"] else "No plot narrative context log available.",
                "tagline": str(row["tagline"]) if row["tagline"] else "",
                "vote_average": float(row["vote_average"]) if row["vote_average"] else 0.0,
                "runtime_minutes": int(row["runtime"]) if row["runtime"] else 0,
                "homepage": str(row["homepage"]) if row["homepage"] else "https://themoviedb.org",
                "director": meta["director"],
                "cast": meta["cast"],
                "similarity_score": round(score, 4)
            })
            
    print("🚀 [TRANSACTION SUCCESSFUL] Retretived and compiled top 10 choices completely.")
    return results

# Overwrite live class instance immediately without resetting any ports or tunnels
PersonalizedVectorSearchEngine.process_personalized_recommendations = process_personalized_recommendations
print("[PERMANENT FIX DEPLOYED] Memory extraction optimization active. Pipeline is blazing fast.")

# ==============================================================================
# KAGGLE CELL 3: CLOUD FIREBASE IDENTITY SERVER INITIALIZATION
# ==============================================================================
import firebase_admin
from firebase_admin import credentials, firestore

if not firebase_admin._apps:
    try:
        # Securely locks onto your real manually verified credential file path
        cred = credentials.Certificate(Settings.SERVICE_ACCOUNT_JSON_PATH)
        firebase_admin.initialize_app(cred)
        print(f"[SUCCESS] Handshake locked onto key target: {Settings.SERVICE_ACCOUNT_JSON_PATH}")
        print("[SUCCESS HANDSHAKE] Kaggle server authenticated against Cloud Firestore cluster.")
    except Exception as e:
        print(f"❌ Real credentials processing failure: {e}")
        print("-> Reverting stack loop fallback to structured mock map configurations...")
        mock_cert_matrix = {
            "type": "service_account", "project_id": "cinema-mesh-auth",
            "private_key_id": "mock_id_12345", "private_key": "-----BEGIN PRIVATE KEY-----\nMOCK\n-----END PRIVATE KEY-----\n",
            "client_email": "firebase-adminsdk@://gserviceaccount.com",
            "client_id": "123456789", "auth_uri": "https://google.com",
            "token_uri": "https://googleapis.com"
        }
        cred = credentials.Certificate(mock_cert_matrix)
        firebase_admin.initialize_app(cred)
        print("[MOCK INTERFACE ACTIVE] Server running with standalone validation constraints.")

# Instantiate core firestore service connector pipeline
db_client = firestore.client()
print("SUCCESS")
print("SUCCESS")
print("SUCCESS")
print("SUCCESS")
print("SUCCESS")
```

```python
# ==============================================================================
# SMART AGENTIC GATEWAY (PART 1 OF 3): SYSTEM ROUTING MODULES & ACL MANIFEST
# ==============================================================================
import time
import json
import os
import base64
import asyncio
import traceback
import hashlib
import uvicorn
import torch
from fastapi import FastAPI, Header, HTTPException, Depends
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from sentence_transformers import SentenceTransformer
from transformers import AutoModelForCausalLM, AutoTokenizer
from cryptography.hazmat.primitives.ciphers.aead import AESGCM

SHARED_PASSPHRASE = "CINE_MESH_SECURE_PHRASE_2026"
DERIVED_KEY_BYTES = hashlib.sha256(SHARED_PASSPHRASE.encode('utf-8')).digest()

DECENTRALIZED_ACL = {
    "client_node_tejas_01": {
        "access_token": "CINE_MESH_R7Q/7925MN",
        "secret_bytes": DERIVED_KEY_BYTES
    }
}

# Persistent In-Memory Session Memory Store for deep conversational context
LOCAL_SESSION_STORE = {}

print("─── OPENING PUBLIC ENTRANCE PROXY REVERSE LINK ───")
get_ipython().system_raw('lt --port 5000 > lt_url.txt &')
time.sleep(3)

try:
    with open("lt_url.txt", "r") as f: 
        print(f"\n🚀 SUCCESS! YOUR FRESH LINK IS READY:\n👉 Live Endpoint URL: {f.read().strip()}\n")
except Exception as url_err:
    print(f"⚠️ Error reading proxy buffer: {url_err}")
# ==============================================================================
# SMART AGENTIC GATEWAY (PART 2 OF 3): GPU NEURAL WEIGHTS ALLOCATION
# ==============================================================================
print("🚀 Mounting embedding weights onto local GPU device configurations...")
cuda_model = SentenceTransformer(Settings.MODEL_NAME, device='cuda')
engine = PersonalizedVectorSearchEngine(Settings.DB_PATH)

print("🤖 Initializing Local Qwen SLM State Engine on T4 GPU device matrix...")
llm_model_name = "Qwen/Qwen2.5-1.5B-Instruct"
llm_tokenizer = AutoTokenizer.from_pretrained(llm_model_name, local_files_only=False)
llm_model = AutoModelForCausalLM.from_pretrained(
    llm_model_name, 
    torch_dtype=torch.float16, 
    device_map="auto"
)

app = FastAPI()
app.add_middleware(
    CORSMiddleware, 
    allow_origins=["*"], 
    allow_credentials=True, 
    allow_methods=["*"], 
    allow_headers=["*"]
)

class EncryptedClientPacket(BaseModel):
    iv: str          
    ciphertext: str  

def api_credential_verifier(x_client_id: str = Header(None), x_access_token: str = Header(None), x_scope_permission: str = Header(None)):
    if not x_client_id or x_client_id not in DECENTRALIZED_ACL:
        raise HTTPException(status_code=403, detail="Invalid credential structure configuration client ID.")
    record = DECENTRALIZED_ACL[x_client_id]
    if record["access_token"] != x_access_token:
        raise HTTPException(status_code=401, detail="Handshake parameters invalidated or token expired.")
    return record

# ==============================================================================
# KAGGLE CELL 4 (PART 3 OF 3): ROLLING UNIFIED CONTEXT SEARCH PIPELINE
# ==============================================================================
@app.post("/api/v3/mesh-chat")
async def process_free_mesh_chat(packet: EncryptedClientPacket, x_scope_permission: str = Header(...), ctx: dict = Depends(api_credential_verifier)):
    print("\n📬 [SMART CONVERSATIONALIST PAYLOAD RECEIVED]")
    try:
        aesgcm = AESGCM(ctx["secret_bytes"])
        try:
            client_iv = base64.b64decode(packet.iv)
            client_ciphertext = base64.b64decode(packet.ciphertext)
            decrypted_raw = aesgcm.decrypt(client_iv, client_ciphertext, None)
            client_json = json.loads(decrypted_raw.decode('utf-8'))
            user_query = client_json["user_query"]
            user_uid = client_json["user_uid"]
            print(f"✅ [DECRYPTION SUCCESS] Query: '{user_query}'")
        except Exception as crypt_err:
            raise HTTPException(status_code=400, detail=f"Cryptographic tracking alignment fault: {str(crypt_err)}")

        if user_uid not in LOCAL_SESSION_STORE:
            LOCAL_SESSION_STORE[user_uid] = {
                "chat_history": [],
                "search_count": 0
            }
            
        session = LOCAL_SESSION_STORE[user_uid]

        # ─── AGENTIC SELF-DETERMINATION SYSTEM SYSTEM INSTRUCTIONS ───
        system_instruction = (
            "You are an incredibly smart, highly capable, and witty movie concierge assistant named CineMesh.\n"
            "INSTRUCTIONS:\n"
            "- Engage in a completely natural, human-like, fluid conversation with the user. You can joke, match their energy, analyze directors, or explore themes.\n"
            "- Your goal is to guide the user to find a great movie. Learn about their genre taste, language preferences, era mood, and style preferences context implicitly through regular conversation.\n"
            "- Never sound like a machine or a static form checklist. Answer any random movie questions they throw at you natively.\n"
            "- THE TRIGGER CAPABILITY: The exact moment you feel you understand their preference mood well enough to give them solid recommendations, or if they explicitly demand choices, you MUST append this exact structural hidden tag to the end of your conversational text reply:\n"
            "  '[INITIATE_VECTOR_SEARCH: <write a dense 1-sentence description combining all extracted user vibes, themes, and genres for mathematical similarity calculations>]'\n"
            "- Example: 'Enjoy your marathon! [INITIATE_VECTOR_SEARCH: dark supernatural english live-action horror movie from the mid 2010s with james wan aesthetic]'\n"
            "- If you are not ready to search yet, simply respond with your regular friendly text conversation sentence, and do NOT output the tag."
        )

        messages = [{"role": "system", "content": system_instruction}]
        for past_chat in session["chat_history"][-10:]:
            messages.append(past_chat)
        messages.append({"role": "user", "content": user_query})

        # Run Local GPU VRAM Inference
        text_inputs = llm_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        model_inputs = llm_tokenizer([text_inputs], return_tensors="pt").to("cuda")
        
        with torch.no_grad():
            generated_ids = llm_model.generate(**model_inputs, max_new_tokens=160, temperature=0.6, do_sample=True)
        generated_ids = [output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)]
        
        decoded_list = llm_tokenizer.batch_decode(generated_ids, skip_special_tokens=True)
        ai_raw_output = decoded_list[0].strip() if (decoded_list and len(decoded_list) > 0) else "I'm processing your thoughts. Tell me more about what kind of film mood we are chasing today!"

        extracted_movies = []
        clean_ai_response = ai_raw_output

        # ─── DUAL-LAYER CONTEXT-AWARE INTENT VALIDATION GATE ───
        has_slm_tag = "[INITIATE_VECTOR_SEARCH:" in ai_raw_output
        has_direct_user_demand = any(w in user_query.lower() for w in ["list", "show cards", "display in cards", "suggest 10", "movies now", "proper list"])

        if has_slm_tag or has_direct_user_demand:
            print("🎯 [UNIFIED PIPELINE TRIGGER] Consolidating complete conversational matrix for search query generation...")
            try:
                # ─── ROLLING FULL CHAT CONTEXT CONCATENATION LAYER ───
                # Extract all raw user input statements from the first message onward
                user_message_history = [m["content"] for m in session["chat_history"] if m["role"] == "user"]
                user_message_history.append(user_query) # Append the current message
                
                # Merge into a single, cohesive, long-form semantic search query string
                unified_search_query = " ".join(user_message_history).strip()

                if has_slm_tag:
                    # If model explicitly generated a summary, append it to reinforce accuracy
                    ai_summary_phrase = ai_raw_output.split("[INITIATE_VECTOR_SEARCH:")[-1].split("]").strip()
                    search_prompt = f"{unified_search_query} {ai_summary_phrase}"
                    clean_ai_response = ai_raw_output.split("[INITIATE_VECTOR_SEARCH:")[0].strip()
                else:
                    search_prompt = unified_search_query
                    clean_ai_response = "Perfect! I've combined your entire conversation timeline choices. Compiling your real database game cards right below:"

                if not clean_ai_response:
                    clean_ai_response = "Processing your global preferences timeline and loading your matching cards below:"

                print(f"🤖 Unified Continuous Search Query String: '{search_prompt}'")
                
                # Execute local SQLite vector similarity metrics on your uncorrupted recovery DB file
                extracted_movies = engine.process_personalized_recommendations(
                    cuda_model, search_prompt, x_scope_permission, [], []
                )
                
                # Clear session history lists to reset memory arrays fresh for the next recommendation session
                session["chat_history"] = []
                session["search_count"] += 1
            except Exception as trigger_err:
                print(f"⚠️ Agentic intercept mapping anomaly: {trigger_err}")
                traceback.print_exc()
        else:
            # Maintain running active memory history registers natively
            session["chat_history"].append({"role": "user", "content": user_query})
            session["chat_history"].append({"role": "assistant", "content": ai_raw_output})

        raw_payload = {"ai_response": clean_ai_response, "movies": extracted_movies, "timestamp": time.time()}
        outbound_nonce = os.urandom(12)
        ciphertext = aesgcm.encrypt(outbound_nonce, json.dumps(raw_payload).encode('utf-8'), None)
        
        print("📤 [OUTBOUND TRANSACTION PREPARED] Returned secure history-consolidated payload matrix.")
        return {
            "iv": base64.b64encode(outbound_nonce).decode('utf-8'), 
            "ciphertext": base64.b64encode(ciphertext).decode('utf-8')
        }
        
    except Exception as e:
        traceback.print_exc()
        raise HTTPException(status_code=500, detail=str(e))

config = uvicorn.Config(app, host="127.0.0.1", port=5000, log_level="warning")
server = uvicorn.Server(config)

print("="*80)
print("🔑 ARCTIC STEP MANIFEST: ROLLING PIPELINE READY")
print("X-Client-ID:       client_node_tejas_01")
print("X-Access-Token:    CINE_MESH_R7Q/7925MN")
print("Symmetric Key:     [SHA-256 AUTOMATIC DERIVATION ACTIVE]")
print("="*80 + "\n")

asyncio.create_task(server.serve())
print("🛰️ Agentic conversational server successfully initialized over active loop.")
```

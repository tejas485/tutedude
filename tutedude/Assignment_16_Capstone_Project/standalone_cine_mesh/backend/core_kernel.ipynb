```python
# ==============================================================================
# MODULE 1: GLOBAL DEPENDENCY CONFIGURATION, PORT PURGING & CHECKUP
# ==============================================================================
import subprocess
import time
import os
import sys
import base64
import json
import re
import hashlib
import asyncio
import sqlite3
import shutil

print("[VALIDATOR_CHECKPOINT] Starting system initialization...")

# Enhanced self-healing socket clearance matrix to free up port 8000
for retry_attempt in range(3):
    try:
        subprocess.run(["fuser", "-k", "8000/tcp"], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
        time.sleep(1)
        check_port = subprocess.run(["lsof", "-t", "-i:8000"], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
        if not check_port.stdout.strip():
            print("[VALIDATOR_CHECKPOINT] Stale port 8000 networks purged cleanly!")
            break
    except Exception:
        pass
    time.sleep(1)

print("[VALIDATOR_CHECKPOINT] Downloading system and python package dependencies...")
subprocess.run(
    "pip install fastapi uvicorn sentence-transformers numpy cryptography pydantic nest-asyncio httpx transformers torch -q",
    shell=True, check=True
)
subprocess.run("npm install -g localtunnel -q", shell=True, check=True)

# Explicit global imports synchronized across cell compilation contexts
import numpy as np
import httpx
import uvicorn
import torch
import nest_asyncio
from fastapi import FastAPI, Header, HTTPException, Depends
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from sentence_transformers import SentenceTransformer
from transformers import AutoModelForCausalLM, AutoTokenizer
from cryptography.hazmat.primitives.ciphers.aead import AESGCM

print("[VALIDATOR_CHECKPOINT] All platform dependencies deployed successfully.")


# ==============================================================================
# MODULE 2: ENVIRONMENT CONFIGURATION & SECURITY REGISTRY
# ==============================================================================
# Initialize loop threading for nested execution states
nest_asyncio.apply()

class Settings:
    # AUTOMATION MOUNT VECTOR: Points directly to your private automated dataset root path
    DB_PATH = "/kaggle/input/tmdb-dataset/tmdb_movie_recommender.db"
    MODEL_NAME = "all-MiniLM-L6-v2"

# SECURE INJECTION: Pulls dynamic passphrase parameters directly from Kaggle Secret environment flags
# Falling back to the default variable if no secure system keys are bound at runtime
SHARED_PASSPHRASE = os.getenv("CINE_MESH_AES_KEY", "CINE_MESH_SECURE_PHRASE_2026")
DERIVED_KEY_BYTES = hashlib.sha256(SHARED_PASSPHRASE.encode('utf-8')).digest()

# Secure client ID allocations linked straight from Flutter compile configurations
CINE_MESH_CLIENT_ID = os.getenv("CINE_MESH_CLIENT_ID", "client_node_tejas_01")
CINE_MESH_ACCESS_TOKEN = os.getenv("CINE_MESH_ACCESS_TOKEN", "CINE_MESH_R7Q/7925MN")

DECENTRALIZED_ACL = {
    CINE_MESH_CLIENT_ID: {
        "access_token": CINE_MESH_ACCESS_TOKEN,
        "secret_bytes": DERIVED_KEY_BYTES
    }
}
LOCAL_SESSION_STORE = {}
print(f"[VALIDATOR_CHECKPOINT] Cryptographic security matrix provisioned safely.")


# ==============================================================================
# MODULE 3: VECTOR SEARCH ENGINE INITIALIZATION (PART 1 OF 2)
# ==============================================================================
class PersonalizedVectorSearchEngine:
    def __init__(self, fallback_db_name: str):
        # Sweeps multiple potential target directories looking for valid database attachments
        possible_paths = [
            Settings.DB_PATH,
            "/kaggle/input/datasets/tejasmali1234/tmdb-dataset",
            os.path.dirname(fallback_db_name)
        ]
        
        detected_db = None
        for base_path in possible_paths:
            if os.path.exists(base_path):
                for root, dirs, files in os.walk(base_path):
                    for f in files:
                        if f.endswith('.db') or f.endswith('.sqlite'):
                            detected_db = os.path.join(root, f)
                            break
                    if detected_db: break
            if detected_db: break
        
        self.source_db = detected_db if detected_db else fallback_db_name
        self.db_file = "/kaggle/working/movie_graph_recovered.db"
        
        print(f"[VALIDATOR_CHECKPOINT] Linking onto verified structural TMDB dataset: {self.source_db}")
        self._execute_live_database_salvage_and_repair()

    def _execute_live_database_salvage_and_repair(self):
        if os.path.exists(self.db_file) and os.path.getsize(self.db_file) > 50000:
            print(f"[VALIDATOR_CHECKPOINT] Production database cache index verified active at: {self.db_file}")
            return

        if self.source_db and os.path.exists(self.source_db) and self.source_db != self.db_file:
            try:
                shutil.copy2(self.source_db, self.db_file)
                print(f"[VALIDATOR_CHECKPOINT] Your TMDB database has been successfully mounted to runtime: {self.db_file}")
            except Exception as e:
                print(f"⚠️ Direct OS file copy blocked ({e}). Running table stream fallback...")
                self._migrate_rows_manually()
        else:
            if os.path.exists(self.db_file):
                print(f"[VALIDATOR_CHECKPOINT] Active database file verified directly inside local workspace path.")
            else:
                print(f"❌ [CRITICAL ERROR] No pre-processed .db file found at path: {self.source_db}")

    def _migrate_rows_manually(self):
        dest_conn = sqlite3.connect(self.db_file)
        dest_c = dest_conn.cursor()
        
        # Fixed: Enhanced structural matrix schema mapping including vector embeddings array cache blobs
        dest_c.execute("""CREATE TABLE IF NOT EXISTS movies (
            id INTEGER PRIMARY KEY, title TEXT, genres TEXT, overview TEXT, tagline TEXT, 
            vote_average REAL, runtime INT, homepage TEXT, director TEXT, cast_members TEXT,
            embedding BLOB)""") # Preserved dense multi-vector embedding payload support
        dest_conn.commit()

        try:
            src_conn = sqlite3.connect(self.source_db)
            src_c = src_conn.cursor()
            
            # Safe checking to verify if source schema contains embedding calculations
            src_c.execute("PRAGMA table_info(movies)")
            columns = [col[1] for col in src_c.fetchall()]
            
            if "embedding" in columns:
                query = "SELECT id, title, genres, overview, tagline, vote_average, runtime, homepage, director, cast_members, embedding FROM movies"
                src_c.execute(query)
                dest_c.executemany("INSERT OR IGNORE INTO movies VALUES (?,?,?,?,?,?,?,?,?,?,?)", src_c.fetchall())
            else:
                query = "SELECT id, title, genres, overview, tagline, vote_average, runtime, homepage, director, cast_members FROM movies"
                src_c.execute(query)
                for row in src_c.fetchall():
                    # Fallback assignment generating zeroed dummy layout vectors for missing entries
                    dest_c.execute("INSERT OR IGNORE INTO movies (id, title, genres, overview, tagline, vote_average, runtime, homepage, director, cast_members, embedding) VALUES (?,?,?,?,?,?,?,?,?,?,?)", row + (None,))
            
            dest_conn.commit()
            src_conn.close()
            print("[VALIDATOR_CHECKPOINT] Your TMDB database has been successfully mounted to runtime: manual_stream")
        except Exception as err:
            print(f"❌ Structural database streaming loop failed: {err}")
        finally:
            dest_conn.close()


# ==============================================================================
# MODULE 4: RECOMMENDATION ENGINE CALCULATIONS (PART 2 OF 2)
# ==============================================================================
def process_personalized_recommendations(self, model, query: str, scope: str, fav_genres: list, liked_movie_ids: list) -> list:
    qv = model.encode(query, convert_to_numpy=True).astype(np.float32)
    results = []
    
    with sqlite3.connect(self.db_file) as conn:
        conn.row_factory = sqlite3.Row
        c = conn.cursor()
        try:
            c.execute("SELECT id, title, overview, tagline, vote_average, runtime, homepage, genres FROM movies")
            backup_rows = c.fetchall()
        except Exception as read_err:
            print(f"❌ Core table extraction constraint error: {read_err}")
            return []
            
        if not backup_rows: return []
        m_ids = []
        text_corpus = []
        movie_metadata_cache = {}
        
        for r in backup_rows:
            current_id = int(r["id"])
            m_ids.append(current_id)
            
            genres_list = [g.strip() for g in str(r["genres"]).split(",") if g.strip()] if r["genres"] else ["Film"]
            genres_joined = "/".join(genres_list)
            
            movie_metadata_cache[current_id] = {
                "row": r, "genres": genres_list, "genres_str": genres_joined, "cast": ["N/A"], "director": "Unknown"
            }
            if scope == "plot":
                text_node = str(r["overview"]) if r["overview"] else str(r["title"])
            else:
                text_node = f"Title: {r['title']}. Genres: {genres_joined}. Tagline: {r['tagline']}."
            text_corpus.append(text_node)
            
        matrix = model.encode(text_corpus, convert_to_numpy=True, batch_size=256, show_progress_bar=False).astype(np.float32)
        dots = np.dot(matrix, qv)
        norms = np.linalg.norm(matrix, axis=1) * (np.linalg.norm(qv) or 1e-9)
        norms[norms == 0] = 1e-9
        base_scores = dots / norms
        
        candidates = []
        for score, m_id in zip(base_scores, m_ids):
            meta = movie_metadata_cache[m_id]
            final_score = float(score)
            genres_lower = meta["genres_str"].lower()
            q_low = query.lower()
            if "bollywood" in q_low or "hindi" in q_low:
                if "bollywood" in genres_lower or "hindi" in genres_lower: final_score += 0.80 
            candidates.append((final_score, m_id))
        
        candidates.sort(key=lambda x: x, reverse=True) 
        for score, m_id in candidates[:10]:
            meta = movie_metadata_cache[m_id]
            row = meta["row"]
            results.append({
                "id": int(m_id), "title": str(row["title"]) if row["title"] else "Unknown Movie",
                "genres": meta["genres"] if meta["genres"] else ["Multi-Genre"],
                "description": str(row["overview"]) if row["overview"] else "No plot details available.",
                "tagline": str(row["tagline"]) if row["tagline"] else "",
                "vote_average": float(row["vote_average"]) if row["vote_average"] else 0.0,
                "runtime_minutes": int(row["runtime"]) if row["runtime"] else 0,
                "homepage": str(row["homepage"]) if row["homepage"] else "https://themoviedb.org",
                "director": meta["director"], "cast": meta["cast"], "similarity_score": round(score, 4)
            })
    return results

PersonalizedVectorSearchEngine.process_personalized_recommendations = process_personalized_recommendations
print("[VALIDATOR_CHECKPOINT] Personalized recommendation algorithm patched successfully.")

# ==============================================================================
# MODULE 5: ZERO-KNOWLEDGE MULTI-TENANT REST ENGINE
# ==============================================================================
class FirestoreUserClient:
    """Handles secure document synchronization directly under individual user tokens."""
    def __init__(self, user_id_token: str, project_id: str = "cinema-mesh-auth"):
        self.token = user_id_token
        # Fixed: Corrected full formal Firestore database target REST collection mapping path
        self.base_url = f"https://googleapis.com{project_id}/databases/(default)/documents"
        self.headers = {
            "Authorization": f"Bearer {user_id_token}", 
            "Content-Type": "application/json"
        }

    def update_user_document(self, user_uid: str, update_fields: dict) -> bool:
        # Construct explicit document target endpoint matching Flutter models
        target_url = f"{self.base_url}/cinema_users/{user_uid}"
        formatted_fields = {k: {"stringValue": str(v)} for k, v in update_fields.items()}
                
        payload = {"fields": formatted_fields}
        mask_params = [("updateMask.fieldPaths", key) for key in update_fields.keys()]
        
        try:
            # Enforce strict client context lifecycle controls preventing socket leaks
            with httpx.Client(timeout=10.0) as client:
                res = client.patch(target_url, headers=self.headers, json=payload, params=mask_params)
                if res.status_code == 200:
                    return True
                else:
                    print(f"⚠️ Firestore sync warning status code: {res.status_code} | Log: {res.text}")
                    return False
        except Exception as sync_fault:
            print(f"❌ Handshake update failed with exception: {sync_fault}")
            return False

print("[VALIDATOR_CHECKPOINT] Zero-Exposure REST Client compilation completed.")


# ==============================================================================
# MODULE 6: LOCALTUNNEL ACTIVATION & CLOUD SYNC
# ==============================================================================
print("[VALIDATOR_CHECKPOINT] Spawning background localtunnel listener proxy routing...")

# Self-Healing Layer: Clear out any duplicate leaking background configurations
try:
    os.system("pkill -f localtunnel")
    os.system("pkill -f lt")
except Exception:
    pass

log_filename = "lt_url.txt"
if os.path.exists(log_filename):
    try: os.remove(log_filename)
    except Exception: pass

# Spawn the background tunnel mapping listener proxy task
subprocess.Popen(f"lt --port 8000 > {log_filename} 2>&1 &", shell=True)

public_tunnel_url = ""
max_polling_attempts = 20
poll_interval_seconds = 1.0

print("[VALIDATOR_CHECKPOINT] Extracting dynamic link address from live log streams...")
for attempt in range(max_polling_attempts):
    time.sleep(poll_interval_seconds)
    if os.path.exists(log_filename):
        try:
            with open(log_filename, "r") as f:
                log_content = f.read().strip()
                # Enhanced robust regex handling variant line outputs natively
                match = re.search(r"https://[a-z0-9\-]+\.loca\.lt", log_content)
                if match:
                    public_tunnel_url = match.group(0)
                    break
                # Fallback matching raw text if printed without explicit delimiters
                elif "url is:" in log_content.lower():
                    extracted = log_content.split("url is:")[-1].strip()
                    if "https://" in extracted:
                        public_tunnel_url = extracted
                        break
        except Exception:
            pass

# 🎯 FIXED: Prints the link clearly in your Kaggle Terminal Output Console
if public_tunnel_url:
    print("\n" + "="*80)
    print(f"🚀 [LIVE LINK ENTRANCE] LOCAL TUNNEL URL: {public_tunnel_url}")
    print("="*80 + "\n")
else:
    print("❌ Anomaly: Local tunnel extraction layer returned a null string profile.")
    if os.path.exists(log_filename):
        with open(log_filename, "r") as f:
            print(f"📋 Raw Localtunnel Logs:\n{f.read()}")



# ==============================================================================
# MODULE 7: MODEL LOADING & FASTAPI GATEWAY SHELL
# ==============================================================================
print("[VALIDATOR_CHECKPOINT] Loading CUDA tokenizers and matrix shapes...")
cuda_model = SentenceTransformer(Settings.MODEL_NAME, device='cuda')
engine = PersonalizedVectorSearchEngine(Settings.DB_PATH)

llm_model_name = "Qwen/Qwen2.5-1.5B-Instruct"
llm_tokenizer = AutoTokenizer.from_pretrained(llm_model_name, local_files_only=False)
llm_model = AutoModelForCausalLM.from_pretrained(
    llm_model_name, torch_dtype=torch.float16, device_map="auto", low_cpu_mem_usage=True
)

app = FastAPI()
app.add_middleware(
    CORSMiddleware, 
    allow_origins=["*"], 
    allow_credentials=True, 
    allow_methods=["*"], 
    allow_headers=["*"]
)

class EncryptedClientPacket(BaseModel):
    iv: str          
    ciphertext: str  

def api_credential_verifier(x_client_id: str = Header(None), x_access_token: str = Header(None)):
    if not x_client_id or x_client_id not in DECENTRALIZED_ACL:
        raise HTTPException(status_code=403, detail="Invalid credential context Client ID.")
    record = DECENTRALIZED_ACL[x_client_id]
    if record["access_token"] != x_access_token:
        raise HTTPException(status_code=401, detail="Handshake parameters invalidated or token expired.")
    return record


# ==============================================================================
# MODULE 8: CONVERSATIONAL INFERENCES ROUTING ENDPOINT
# ==============================================================================
@app.post("/api/v3/mesh-chat")
async def process_free_mesh_chat(
    packet: EncryptedClientPacket, 
    x_firebase_user_token: str = Header(...),
    x_scope_permission: str = Header("plot"),
    ctx: dict = Depends(api_credential_verifier)
):
    try:
        aesgcm = AESGCM(ctx["secret_bytes"])
        client_iv = base64.b64decode(packet.iv)
        client_ciphertext = base64.b64decode(packet.ciphertext)
        
        decrypted_raw = aesgcm.decrypt(client_iv, client_ciphertext, None)
        client_json = json.loads(decrypted_raw.decode('utf-8'))
        user_query = client_json["user_query"]
        user_uid = client_json["user_uid"]

        if user_uid not in LOCAL_SESSION_STORE:
            LOCAL_SESSION_STORE[user_uid] = {"chat_history": [], "search_count": 0}
        session = LOCAL_SESSION_STORE[user_uid]

        system_instruction = (
            "You are an incredibly smart, highly capable, and witty movie concierge assistant named CineMesh.\n"
            "INSTRUCTIONS:\n"
            "- Engage in a completely natural, human-like, fluid conversation with the user.\n"
            "- THE TRIGGER CAPABILITY: The exact moment you feel you understand their preference mood well enough to give them solid recommendations, append this exact structural hidden tag to the end of your conversational text reply:\n"
            "  '[INITIATE_VECTOR_SEARCH: <write a dense 1-sentence description combining all extracted user vibes, themes, and genres for mathematical similarity calculations>]'\n"
            "- If you are not ready to search yet, simply respond with your regular friendly text conversation sentence, and do NOT output the tag."
        )

        messages = [{"role": "system", "content": system_instruction}]
        for past_chat in session["chat_history"][-8:]: # Capped sliding context window tracking
            messages.append(past_chat)
        messages.append({"role": "user", "content": user_query})

        text_inputs = llm_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        model_inputs = llm_tokenizer([text_inputs], return_tensors="pt", padding=True).to("cuda")
        
        with torch.no_grad():
            generated_ids = llm_model.generate(
                input_ids=model_inputs.input_ids,
                attention_mask=model_inputs.attention_mask, # Fixed: Enforced explicit attention masking
                max_new_tokens=160, 
                temperature=0.6, 
                do_sample=True,
                pad_token_id=llm_tokenizer.eos_token_id
            )
            
        generated_ids = [output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)]
        decoded_list = llm_tokenizer.batch_decode(generated_ids, skip_special_tokens=True)
        ai_raw_output = decoded_list[0].strip() if (decoded_list and len(decoded_list) > 0) else "Tell me more about what kind of film mood we are chasing today!"

        extracted_movies = []
        clean_ai_response = ai_raw_output

        has_slm_tag = "[INITIATE_VECTOR_SEARCH:" in ai_raw_output
        has_direct_user_demand = any(w in user_query.lower() for w in ["list", "show cards", "display in cards", "suggest 10", "movies now"])

        if has_slm_tag or has_direct_user_demand:
            try:
                # Optimization: Slice history length to prevent infinite context bloat
                sliced_history = [m["content"] for m in session["chat_history"] if m["role"] == "user"][-4:]
                sliced_history.append(user_query)
                unified_search_query = " ".join(sliced_history).strip()

                if has_slm_tag:
                    split_chunks = ai_raw_output.split("[INITIATE_VECTOR_SEARCH:")
                    ai_summary_phrase = split_chunks[-1].split("]")[0].strip()
                    search_prompt = f"{unified_search_query} {ai_summary_phrase}"
                    clean_ai_response = split_chunks[0].strip()
                else:
                    search_prompt = unified_search_query
                    clean_ai_response = "Perfect! I've combined your entire conversation timeline choices. Compiling your real database game cards right below:"

                extracted_movies = engine.process_personalized_recommendations(cuda_model, search_prompt, x_scope_permission, [], [])
                session["chat_history"] = []
                session["search_count"] += 1
            except Exception as search_fault:
                print(f"⚠️ Vector fallback intercept trace: {search_fault}")
        else:
            session["chat_history"].append({"role": "user", "content": user_query})
            session["chat_history"].append({"role": "assistant", "content": ai_raw_output})

        # Securely sync dynamic tunnel URLs asynchronously away from main thread
        if public_tunnel_url:
            try:
                rest_client = FirestoreUserClient(x_firebase_user_token)
                # Offloaded connection steps asynchronously preventing thread blocks
                loop = asyncio.get_event_loop()
                await loop.run_in_executor(
                    None, 
                    rest_client.update_user_document, 
                    user_uid, 
                    {"current_tunnel_url": public_tunnel_url}
                )
            except Exception as sync_err:
                print(f"⚠️ Non-blocking token cloud mapping bypass trace: {sync_err}")

        raw_payload = {"ai_response": clean_ai_response, "movies": extracted_movies, "timestamp": time.time()}
        outbound_nonce = os.urandom(12)
        ciphertext = aesgcm.encrypt(outbound_nonce, json.dumps(raw_payload).encode('utf-8'), None)
        
        return {
            "iv": base64.b64encode(outbound_nonce).decode('utf-8'), 
            "ciphertext": base64.b64encode(ciphertext).decode('utf-8')
        }
    except Exception as general_fault:
        print(f"❌ Core API pipeline crash intercepted: {general_fault}")
        raise HTTPException(status_code=500, detail=str(general_fault))

# Bind onto localhost port 8000 headlessly
config = uvicorn.Config(app, host="127.0.0.1", port=8000, log_level="warning")
server = uvicorn.Server(config)

print("\n" + "="*80)
print("[VALIDATOR_CHECKPOINT] RUNTIME GATEWAY SECURED: ENGINE FULLY DEPLOYED")
print("Target Port Mapping: Port 8000")
print("="*80 + "\n")

async def run_server_headless():
    await server.serve()

asyncio.get_event_loop().create_task(run_server_headless())
print("[VALIDATOR_CHECKPOINT] Server engine background thread initialized cleanly.")




```
